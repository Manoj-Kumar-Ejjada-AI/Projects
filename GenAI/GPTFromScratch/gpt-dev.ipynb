## Read the data

```python
with open('input.txt', 'r', encoding = 'utf-8') as f:
  text = f.read()
```

## Checking the length of dataset by characters

```python
print(len(text))
```

## Printing 1st 1000 characters

```python
print(text[:1000])
```

## Checking the all possible characters in the dataset

```python
chars = sorted(list(set(text)))
vocab_size = len(chars)
print(''.join(chars))
print(vocab_size)
```

### Encode and Decode

```python
stoi = {char:i for i,char in enumerate(chars)}
itos = {i:char for i,char in enumerate(chars)}
encode = lambda s: [stoi[c] for c in s]
decode = lambda n: ''.join([itos[num] for num in n])
print(stoi)
print(itos)
print(encode("Hi! How are you?"))
print(decode([21,1,39,51,1,44,47,52,43,8]))
print(decode(encode("This is Manoj")))
```

### Encode entire dataset and store it in `torch.tensor`

```python
import torch
data = torch.tensor(encode(text), dtype = torch.long)
print(data.shape, data.dtype)
print(data[:1000])
```

### split data into train and target data

```python
n = int(0.9*(len(data)))
train_data = data[:n]
val_data = data[n:]
```

```python
block_size = 8
train_data[:block_size+1]
```

```python
for text in range(block_size):
  train = train_data[:text+1]
  val = train_data[text+1]
  print("If input is ",train, " output is ",val)
```

```python
torch.manual_seed(1337)
batch_size = 4
block_size = 8

def get_batch(split):
  data = train_data if split == "train" else val_data
  ix = torch.randint(len(data) - block_size, (batch_size,))
  # print(ix)
  x = torch.stack([data[i:i+block_size] for i in ix])
  # print(x)
  y = torch.stack([data[i+1:i+block_size+1] for i in ix])
  return x,y

xa,xb = get_batch("train")
print("inputs:")
print(xa.shape)
print(xa)
print("outputs:")
print(xb.shape)
print(xb)
print("-"*50)

for b in range(batch_size):  # Batch dimension (B)
  for t in range(block_size): # Time dimension (T)
    train = xa[b,:t+1]
    pred = xb[b,t]
    print(f"when input is {train} output is {pred}")
```
